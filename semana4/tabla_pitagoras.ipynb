Este programa construye la tabla de Pitágoras (tabla de multiplicar) como una matriz de 10 × 10 usando listas anidadas y ciclos `for`.
Después, la función `imprimir_tabla` recorre la matriz y la muestra en pantalla en forma de tabla, con cada número separado por un tabulador.
La función `consultar_producto` regresa el producto de dos factores consultando la matriz, y el programa principal pide el renglón y la columna al usuario, valida que ambos estén entre 1 y 10 y muestra el resultado.

## Parte A: la matriz

```python
n = 10  # tamaño de la tabla (empieza con 3 para probar)

tabla = []
for num_fila in range(1, n + 1):
    fila = []
    for num_columna in range(1, n + 1):
        fila.append(num_fila * num_columna)
    tabla.append(fila)
```

```python
# Comprobaciones con 10 x 10
print(tabla[6][7])  # debe ser 56 (7 x 8)
print(tabla[9][9])  # debe ser 100 (10 x 10)
```

## Parte B: imprimir_tabla

```python
def imprimir_tabla(tabla):
    for fila in tabla:
        for n in fila:
            print(n, end="\t")
        print()


imprimir_tabla(tabla)
```

## Parte C: consultar_producto

```python
def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]


resultado = consultar_producto(tabla, 7, 8)
print(resultado)

resultado = consultar_producto(tabla, 1, 1)
print(resultado)

resultado = consultar_producto(tabla, 10, 10)
print(resultado)
```

## Parte D: programa principal

```python
renglon = int(input("Renglón (1 a 10): "))
while renglon < 1 or renglon > 10:
    print("El renglón debe estar entre 1 y 10.")
    renglon = int(input("Renglón (1 a 10): "))

columna = int(input("Columna (1 a 10): "))
while columna < 1 or columna > 10:
    print("La columna debe estar entre 1 y 10.")
    columna = int(input("Columna (1 a 10): "))

producto = consultar_producto(tabla, renglon, columna)
print(f"{renglon} x {columna} = {producto}")
```
