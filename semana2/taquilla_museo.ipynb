Este programa atiende visitantes en la taquilla de un museo. Calcula el precio según la edad, aplica un solo descuento por boleto y, al escribir `fin`, muestra el corte de caja con los visitantes cobrados, el total en descuentos y el total a pagar.

```python
PRECIO_MENOR = 30
PRECIO_ADULTO = 45
DESC_ADULTO_MAYOR = 0.12
DESC_ESTUDIANTE_PROFESOR = 0.10

visitantes = 0
total_descuentos = 0
total_pagar = 0

print("TAQUILLA DEL MUSEO")

while True:
    dato = input("Edad del visitante (o 'fin' para cerrar la taquilla): ")

    if dato == "fin":
        break

    edad = int(dato)

    if edad < 0 or edad > 110:
        print("Edad no válida, no se cobra.")
        continue

    if edad < 3:
        base = 0
        descuento = 0
        tipo_texto = "Menor"
    elif edad < 18:
        base = PRECIO_MENOR
        descuento = 0
        tipo_texto = "Menor"
    else:
        base = PRECIO_ADULTO
        tipo = int(input("Tipo: 1) Estudiante  2) Profesor  3) Ninguno: "))
        if edad >= 60:
            descuento = base * DESC_ADULTO_MAYOR
            tipo_texto = "Adulto mayor"
        elif tipo == 1:
            descuento = base * DESC_ESTUDIANTE_PROFESOR
            tipo_texto = "Estudiante"
        elif tipo == 2:
            descuento = base * DESC_ESTUDIANTE_PROFESOR
            tipo_texto = "Profesor"
        else:
            descuento = 0
            tipo_texto = "Ninguno"

    paga = base - descuento
    visitantes = visitantes + 1
    total_descuentos = total_descuentos + descuento
    total_pagar = total_pagar + paga

    print(f"Visitante {visitantes} | Edad: {edad} | {tipo_texto} | Base: ${base:.2f} | Descuento: ${descuento:.2f} | A pagar: ${paga:.2f}")

print("CORTE DE CAJA")
print(f"Visitantes cobrados: {visitantes}")
print(f"Total en descuentos: ${total_descuentos:.2f}")
print(f"TOTAL A PAGAR: ${total_pagar:.2f}")

```
