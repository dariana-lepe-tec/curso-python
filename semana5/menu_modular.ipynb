Este programa es un menú que llama a una función por cada sección: tuplas, diccionarios, excepciones y strings.
El menú principal (opciones 1 a 5) repite hasta que se elige Salir. Cada opción llama a su función y la opción inexistente muestra "Opción inválida". Al final del notebook están las pruebas ejecutadas y la tabla de evidencia.

```python
# ---------- Parte A: Tuplas ----------

def sumar_tupla(tupla):
    """Recorre la tupla, acumula la suma y la regresa."""
    total = 0                 # el acumulador empieza en 0, antes del ciclo
    for numero in tupla:
        total = total + numero
    return total


def seccion_tuplas():
    print("=== SECCIÓN TUPLAS ===")

    # 1. Tupla original con 5 números
    numeros = (4, 8, 15, 16, 23)
    print("Tupla original:", numeros)

    # 2. Tercer elemento (índice 2, porque se cuenta desde 0)
    print("Tercer elemento:", numeros[2])

    # 3. Pedir dos números
    a = int(input("Escribe el primer número: "))
    b = int(input("Escribe el segundo número: "))

    # 4. Tupla nueva: la original NO se modifica
    nueva_tupla = numeros + (a, b)
    print("Tupla nueva:", nueva_tupla)
    print("Tupla original (sin cambios):", numeros)

    # 5. Convertir a lista, ordenar y mostrar
    lista = list(nueva_tupla)
    lista.sort()
    print("Lista ordenada:", lista)

    # 6. Sumar con la función auxiliar
    suma = sumar_tupla(nueva_tupla)
    print("Suma de la tupla nueva:", suma)
```

```python
# ---------- Parte B: Diccionarios ----------

def buscar_telefono(contactos, nombre):
    """Regresa el teléfono del contacto, o None si no existe (sin KeyError)."""
    return contactos.get(nombre, None)


def seccion_diccionarios():
    print("=== SECCIÓN DICCIONARIOS ===")

    # 1. Diccionario con 3 contactos: clave = nombre, valor = teléfono
    contactos = {
        "Ana": "442-123-4567",
        "Luis": "442-765-4321",
        "María": "442-555-1234",
    }

    # 2. Agregar un contacto nuevo
    nombre = input("Nombre del nuevo contacto: ").strip().title()
    telefono = input("Teléfono del nuevo contacto: ").strip()
    contactos[nombre] = telefono

    # 3. Mostrar solo los nombres
    print("Tus contactos:")
    for clave in contactos.keys():
        print("-", clave)

    # 4. Buscar un contacto
    buscado = input("¿A quién quieres buscar? ").strip().title()
    resultado = buscar_telefono(contactos, buscado)

    # 5. Mostrar el teléfono o el mensaje
    if resultado is not None:
        print(f"El teléfono de {buscado} es {resultado}")
    else:
        print(f"{buscado} no está en tus contactos")
```

```python
# ---------- Parte C: Excepciones ----------

def seccion_excepciones():
    print("=== SECCIÓN EXCEPCIONES ===")
    try:
        a = int(input("Primer número entero: "))
        b = int(input("Segundo número entero: "))
        # Primero se calcula todo; si b es 0, falla aquí y no se imprime nada a medias
        suma = a + b
        division = a / b
        print("Suma:", suma)
        print("División:", division)
    except ValueError:
        print("Error: debes escribir un número entero (no letras ni dejarlo vacío).")
    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero. Intenta con otro segundo número.")
```

```python
# ---------- Parte D: Strings ----------

def contar_palabras(texto):
    """Regresa la cantidad de palabras del texto (no imprime)."""
    palabras = texto.split()
    return len(palabras)


def seccion_strings():
    print("=== SECCIÓN STRINGS ===")
    mensaje = "el metal nunca muere"
    print("Mensaje:", mensaje)

    # 1. Longitud
    print("Longitud:", len(mensaje))

    # 2. Mayúsculas (upper() regresa un texto nuevo; mensaje no cambia)
    print("En mayúsculas:", mensaje.upper())

    # 3. Reemplazar una palabra por otra
    reemplazado = mensaje.replace("metal", "rock")
    print("Con replace():", reemplazado)

    # 4. Contar palabras con la función auxiliar
    total_palabras = contar_palabras(mensaje)
    print("Número de palabras:", total_palabras)
```

```python
# ---------- Parte E: Menú ----------

while True:
    print("\n1) Tuplas  2) Diccionarios  3) Excepciones  4) Strings  5) Salir")
    opcion = input("Elige una opción: ")

    if opcion == "1":
        seccion_tuplas()
    elif opcion == "2":
        seccion_diccionarios()
    elif opcion == "3":
        seccion_excepciones()
    elif opcion == "4":
        seccion_strings()
    elif opcion == "5":
        print("¡Hasta luego!")
        break
    else:
        print("Opción inválida")
```
