### Ejercicio 9: Descuento escalonado
**Enunciado:** Pedir el monto de una compra y aplicar un descuento según el rango: menos de `$100` no hay descuento, de `$100` a `$499` se descuenta 10% y de `$500` en adelante, 20%. Mostrar el precio final con dos decimales.

| Entrada | Proceso | Salida |
|---|---|---|
| `monto` (float) | Elegir el porcentaje con `if / elif / else`; `descuento = monto * porcentaje`; `precio_final = monto - descuento` | Precio final con 2 decimales |

**Procedimiento:** Leí el monto con `float()` para aceptar centavos. Revisé los rangos empezando por el más alto (`>= 500`), así el `elif monto >= 100` ya no necesita la condición `monto < 500`, porque si llegó ahí es que no fue mayor o igual a 500. El `else` cubre las compras menores a `$100` con 0% de descuento. Al final imprimí con `:.2f` para mostrar siempre dos decimales.

```python
monto = float(input("Ingresa el monto de la compra: "))

# Elegir el porcentaje, empezando por el rango más alto
if monto >= 500:
    porcentaje = 0.20
elif monto >= 100:
    porcentaje = 0.10
else:
    porcentaje = 0

descuento = monto * porcentaje
precio_final = monto - descuento

print(f"Precio final con descuento: ${precio_final:.2f}")
```

### Ejercicio 17: Factorial con for
**Enunciado:** Pedir un entero positivo `n` y calcular su factorial, es decir, el producto de todos los enteros de 1 hasta `n`, usando un ciclo `for`.

| Entrada | Proceso | Salida |
|---|---|---|
| `n` (int) | `factorial = 1`; para `i` de 1 a `n`: `factorial = factorial * i` | El factorial de `n` |

**Procedimiento:** Usé un acumulador de producto que empieza en **1** y no en 0, porque cualquier número multiplicado por 0 da 0 y anularía el resultado. El `range(1, n + 1)` recorre de 1 hasta `n` incluido (el límite superior de `range` no se incluye, por eso el `+ 1`). Si `n` es 0, el ciclo no entra y el resultado queda en 1, que es justo el valor del factorial de 0.

```python
n = int(input("Ingresa un entero positivo: "))

factorial = 1          # acumulador de producto: empieza en 1, no en 0
for i in range(1, n + 1):
    factorial = factorial * i

print(f"El factorial de {n} es: {factorial}")
```

### Ejercicio 13: Promedio con centinela -1
**Enunciado:** Capturar calificaciones una por una hasta que el usuario escriba `-1` (valor centinela) y mostrar el promedio de las calificaciones capturadas.

| Entrada | Proceso | Salida |
|---|---|---|
| Calificaciones (float) hasta `-1` | Acumular `suma` y `contador` dentro de un `while True`; salir con `break` al leer `-1`; `promedio = suma / contador` | El promedio, o un aviso si no hubo datos |

**Procedimiento:** Inicialicé `suma` y `contador` en 0 antes del ciclo. Usé `while True` porque no sé cuántas calificaciones habrá; el `-1` se revisa **primero**, antes de sumar, para que el centinela no se cuente como calificación. Al salir, verifico que `contador` no sea 0 para evitar un `ZeroDivisionError` si el usuario escribe `-1` desde el inicio.

```python
suma = 0
contador = 0

while True:
    cal = float(input("Calificación (-1 para terminar): "))
    if cal == -1:          # el centinela se revisa antes de sumar
        break
    suma = suma + cal
    contador = contador + 1

if contador == 0:
    print("No se capturaron calificaciones")
else:
    promedio = suma / contador
    print("El promedio es:", promedio)
```

### Ejercicio 23: Promedio con validación de rango
**Enunciado:** Igual que el ejercicio 13, pero si una calificación está fuera del rango de 0 a 10 se muestra un mensaje de error y esa calificación no se toma en cuenta para el promedio.

| Entrada | Proceso | Salida |
|---|---|---|
| Calificaciones (float) hasta `-1` | Revisar el centinela; si `cal < 0 or cal > 10`, mostrar error y `continue`; acumular solo las válidas | Mensaje de error por cada inválida y el promedio de las válidas |

**Procedimiento:** Partí de la solución del ejercicio 13. Después de revisar el `-1` agregué la validación del rango: si la calificación es menor que 0 o mayor que 10, imprimo el error y uso `continue`, que regresa al inicio del ciclo **sin** ejecutar la suma ni el conteo. Así la calificación inválida no afecta el promedio. El orden importa: si validara el rango antes del `-1`, el centinela se marcaría como error y el programa nunca terminaría.

```python
suma = 0
contador = 0

while True:
    cal = float(input("Calificación (-1 para terminar): "))
    if cal == -1:
        break
    if cal < 0 or cal > 10:    # validación de rango
        print("Error: calificación fuera de rango (0-10)")
        continue               # vuelve a pedir sin sumar ni contar
    suma = suma + cal
    contador = contador + 1

if contador == 0:
    print("No se capturaron calificaciones válidas")
else:
    promedio = suma / contador
    print("El promedio es:", promedio)
```

### Ejercicio 24: Minicalculadora con menú
**Enunciado:** Mostrar un menú con las opciones `1` suma, `2` resta, `3` multiplicación y `4` salir. Para cada operación se piden dos números y se muestra el resultado; el menú se repite hasta que el usuario elija salir.

| Entrada | Proceso | Salida |
|---|---|---|
| `opcion` (texto) y dos números (float) | Ciclo `while True`; `break` con la opción 4; `if / elif` para elegir la operación; `else` para opciones no válidas | Resultado de cada operación y "Fin del programa" |

**Procedimiento:** Usé `while True` para mostrar el menú en cada vuelta. La opción se lee como texto (sin convertir) y se compara con `"1"`, `"2"`, etc., así una entrada como `"x"` no truena el programa. La opción `"4"` hace `break`. Los dos números solo se piden si la opción es una operación válida (`in ("1", "2", "3")`); cualquier otra opción cae en el `else` con el aviso. El mensaje "Fin del programa" va fuera del ciclo, para que se imprima una sola vez al salir.

```python
while True:
    print("\n--- MENÚ ---")
    print("1. Suma")
    print("2. Resta")
    print("3. Multiplicación")
    print("4. Salir")
    opcion = input("Elige una opción: ")

    if opcion == "4":
        break
    elif opcion in ("1", "2", "3"):
        a = float(input("Primer número: "))
        b = float(input("Segundo número: "))
        if opcion == "1":
            print("Suma:", a + b)
        elif opcion == "2":
            print("Resta:", a - b)
        else:
            print("Multiplicación:", a * b)
    else:
        print("Opción no válida")

print("Fin del programa")
```
